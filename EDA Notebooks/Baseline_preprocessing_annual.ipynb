```python
import pandas as pd
```

```python
data = pd.read_csv("C:/Users/katik/Downloads/baseline_annual.csv")
```

```python
# Setting options to display all rows and columns
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Print the full list of columns
print(data.columns.tolist())
```

```python
data = data.drop_duplicates()
```

```python
import numpy as np
fill_vals = [-9999, 9999, -9999.0, 9999.0, '', 'N/A', 'nan'] 
data = data.replace(fill_vals, np.nan)
```

```python
print(data.isnull().sum().sort_values(ascending=False).head(20))  # Top 20 columns with most nulls
```

```python
# Drop columns with >80% missing
drop_threshold = 0.80
cols_to_drop = [col for col in data.columns if data[col].isnull().mean() > drop_threshold]
print("Will drop:", cols_to_drop)
data = data.drop(cols_to_drop, axis=1)
```

```python
print(data.shape)
print(data.isnull().sum().sort_values(ascending=False).head(20))
```

```python
# Numeric columns
num_cols = data.select_dtypes(include=['number']).columns
data[num_cols] = data[num_cols].fillna(data[num_cols].median())

# Categorical columns
cat_cols = data.select_dtypes(include='object').columns
for col in cat_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

```python
total_nulls = data.isnull().sum().sum()
print("Total nulls left:", total_nulls)  # Should be 0
```

```python
print("Duplicate rows:", data.duplicated().sum())
# Optionally remove:
# data = data.drop_duplicates()
```

```python
print(data.describe(include='all').T)  
```
