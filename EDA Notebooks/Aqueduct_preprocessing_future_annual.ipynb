```python
import pandas as pd
import re
data1 = pd.read_csv("C:/Users/katik/Downloads/CSV/CSV/Aqueduct40_future_annual_y2023m07d05.csv")




# Define mapping dictionaries and parser function
scenario_map = {
    "bau": "Business as Usual",
    "opt": "Optimistic",
    "pes": "Pessimistic"
}
year_map = {
    "30": "2030 (2015-2045)",
    "50": "2050 (2035-2065)",
    "80": "2080 (2065-2095)"
}
var_map = {
    "ba": "Blue Water Availability",
    "ww": "Withdrawals",
    "ws": "Water Stress",
    "wd": "Water Depletion",
    "iv": "Interannual Variability",
    "sv": "Seasonal Variability"
}
type_map = {
    "r": "Raw Value (Median of 5 GCMs)",
    "s": "Score [0-5 scale]",
    "l": "Label (Threshold Explanation)",
    "c": "Category (Integer for Visuals)"
}

def get_full_col(col):
    if col in ["fid", "pfaf_id"]:
        return {"fid": "Feature ID", "pfaf_id": "Pfafstetter ID"}[col]
    match = re.match(r'(bau|opt|pes)(30|50|80)_([a-z]{2})_x_([rslc])', col)
    if match:
        scenario, year, var, ctype = match.groups()
        scenario_full = scenario_map.get(scenario, scenario)
        year_full = year_map.get(year, year)
        var_full = var_map.get(var, var)
        type_full = type_map.get(ctype, ctype)
        return f"{var_full} ({scenario_full}, Year {year_full}) - {type_full}"
    return col

# Build column conversion dict
col_fullforms = {col: get_full_col(col) for col in data1.columns}

# Rename columns
data1.rename(columns=col_fullforms, inplace=True)

# Save to CSV
data1.to_csv("Aqueduct_future.csv", index=False)
```

```python
#Clean & Standardize Column Names
import re
data1.columns = [
    re.sub(r'[^A-Z0-9_]', '_', col.upper().replace(' ', '_'))
    for col in data1.columns
]
```

```python
 #Identify and Replace Fill/Missing Values
import numpy as np
data1.replace([-9999, -9999.0], np.nan, inplace=True)
```

```python
# Summarize Nulls and Cardinality
# Null counts
print(data1.isnull().sum().sort_values(ascending=False).head(20))

# High cardinality columns
print("Columns with >95% unique:", [col for col in data1.columns if data1[col].nunique()/len(data1) > 0.95])
```

```python
# Numeric columns
num_cols = data1.select_dtypes(include=['number']).columns
data1[num_cols] = data1[num_cols].fillna(data1[num_cols].median())
# Categorical/object columns
cat_cols = data1.select_dtypes(include='object').columns
for col in cat_cols:
    data1[col] = data1[col].fillna(data1[col].mode().iloc[0])
```
